```python
import pandas as pd
import os
```

```python

# Function to split large DataFrame into multiple smaller DataFrames
def split_dataframe(df, max_rows=1048575):
    # Split the DataFrame into chunks of 'max_rows' each
    chunks = []
    for start in range(0, len(df), max_rows):
        chunks.append(df.iloc[start:start + max_rows])
    return chunks

# Example to convert CSV to Excel, splitting if necessary
csv_file = 'filtered_trains.csv'  # Change to your CSV file path
excel_file = 'filtered_trains_LeipzigHbf_2024_07.xlsx'  # Change to desired output Excel file

# Load the CSV file
df = pd.read_csv(csv_file)

# Split DataFrame if it's too large
chunks = split_dataframe(df)

# Use openpyxl as the engine for saving to Excel
with pd.ExcelWriter(excel_file, engine='openpyxl') as writer:
    for idx, chunk in enumerate(chunks):
        chunk.to_excel(writer, sheet_name=f'Sheet_{idx+1}', index=False)

print(f"CSV file '{csv_file}' successfully converted to Excel file '{excel_file}'.")
```
