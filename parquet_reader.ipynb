```python
#parquet_file = 'data-2054-02.parquet'
# Read the Parquet file using the pyarrow engine
df = pd.read_parquet('data-2024-07.parquet')    
```

```python
# Read the Parquet file using the pyarrow engine
#df = pd.read_parquet(parquet_file, engine='pyarrow')

# Define the station you are interested in
station_name = "Stuttgart Hbf"  # Change this to your target station

# Filter the DataFrame to get only rows where the station matches the desired station
df_filtered = df[df["station"] == station_name]

# Exclude bus entries to keep only actual trains
#df_trains = df_filtered[~df_filtered["train_type"].str.contains("Bus", na=False)]

# Further filter by specific train types (e.g., RE, ICE, ME)
selected_train_types = ["RE", "ICE", "ME", "RB"]
df_trains = df_filtered[df_filtered["train_type"].isin(selected_train_types)]

required_columns = [
    "arrival_planned_time", "arrival_change_time", 
    "departure_planned_time", "departure_change_time", 
    "delay_in_min", "time"
]

# Ensure all required columns exist before filtering
if all(col in df.columns for col in required_columns):
    # Drop rows where any of the required columns have NaN values
    df_trains = df_trains.dropna(subset=required_columns)
# Further filter by final destination station
#final_destination = ["Würzburg Hbf", "Jena"]  # Change this to your desired final destination
#df_trains = df_trains[df_trains["final_destination_station"].isin(final_destination)]

# Select only the first 100 rows
#df_first_100 = df_trains.head(100)

# Save the filtered data to a CSV file
df_trains.to_csv('filtered_trains.csv', index=False)
print("Filtered data has been saved to 'filtered_trains.csv'.")
```
